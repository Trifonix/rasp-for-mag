# Практикум 2.2 — Геометрия и градиентный спуск

```
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

## Задание 1: Геометрия (dot, угол, поворот)

```
a = np.array([1., 0.])
b = np.array([0.5, np.sqrt(3)/2])
dot = np.dot(a, b)
cos_theta = dot / (np.linalg.norm(a) * np.linalg.norm(b))
theta = np.degrees(np.arccos(np.clip(cos_theta, -1, 1)))
phi = np.radians(30)
R = np.array([[np.cos(phi), -np.sin(phi)], [np.sin(phi), np.cos(phi)]])
a_rot = R @ a
print(f'dot={dot:.3f}, theta≈{theta:.1f}°')
print('a_rot:', a_rot)
```

## Задание 2: Градиентный спуск 1D

```
def f(x):
    return (x-3)**2

x = 8.0
lr = 0.2
xs, fs = [], []
for step in range(25):
    grad = 2*(x-3)
    x -= lr*grad
    xs.append(x)
    fs.append(f(x))
    print(f'{step:02d}: x={x:.4f}, f(x)={f(x):.6f}')
plt.figure(); plt.plot(fs); plt.title('Сходимость f(x)'); plt.xlabel('step'); plt.ylabel('f(x)'); plt.grid(True)
```

## Задание 3: Градиентный спуск 2D

```
def f2(w):
    return (w[0]-1)**2 + 2*(w[1]+2)**2

w = np.array([3.0, -6.0])
lr = 0.1
vals = []
for step in range(30):
    grad = np.array([2*(w[0]-1), 4*(w[1]+2)])
    w -= lr * grad
    vals.append(f2(w))
    print(f'{step:02d}: w={w}, f(w)={f2(w):.6f}')
plt.figure(); plt.plot(vals); plt.title('Сходимость f(w)'); plt.xlabel('step'); plt.ylabel('f(w)'); plt.grid(True)
```
