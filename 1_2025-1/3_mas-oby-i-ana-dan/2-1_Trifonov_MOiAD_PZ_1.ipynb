Используем стандартные библиотеки анализа данных и визуализации.

```python
!pip install pandas matplotlib numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Шаг 2. Создание и сохранение CSV-файла

Для демонстрации создадим простой CSV-файл с данными о продажах.

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [120, 150, 170, 160, 180, 200]
}

df = pd.DataFrame(data)
df.to_csv('sales_data.csv', index=False)

df
```

## Шаг 3. Чтение CSV-файла

Считываем данные из файла с помощью pandas.

```python
df = pd.read_csv('sales_data.csv')
df
```

## Шаг 4. Линейный график (matplotlib)

Построим линейный график изменения продаж по месяцам.

```python
plt.figure(figsize=(8, 5))
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title('Продажи по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Продажи')
plt.grid(True)
plt.show()
```

## Шаг 5. Столбчатый график

Построим столбчатую диаграмму для тех же данных.

```python
plt.figure(figsize=(8, 5))
plt.bar(df['Month'], df['Sales'])
plt.title('Столбчатая диаграмма продаж')
plt.xlabel('Месяц')
plt.ylabel('Продажи')
plt.show()
```

## Шаг 6. Сравнение Jupyter Notebook и .py-скрипта

**Jupyter Notebook:**
- Удобная работа с ячейками
- Немедленная визуализация результатов
- Подходит для анализа и обучения

**Python-скрипт (.py):**
- Удобен для автоматизации и запуска из терминала
- Меньше визуальных возможностей
- Лучше подходит для production-кода
