## Подготовка данных (Data Preparation)


```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

VARIANT = 1
data_raw = pd.read_csv(f"PZ2_variant_{VARIANT}.csv", encoding="utf-8-sig", sep=";")
data = data_raw.copy()
data.head()
```

## Анализ качества данных

```
data.isnull().sum(), data.duplicated().sum()
```

## Удаление дубликатов

```
data = data.drop_duplicates()
```

## Обработка пропусков

```

for c in data.select_dtypes(include=[np.number]).columns:
    data[c] = data[c].fillna(data[c].median())
data.isnull().sum()
```

## Обработка выбросов (IQR)

```

def iqr_clean(df):
    res=df.copy()
    for c in res.select_dtypes(include=[np.number]).columns:
        q1,q3=res[c].quantile([0.25,0.75])
        iqr=q3-q1
        res=res[(res[c]>=q1-1.5*iqr)&(res[c]<=q3+1.5*iqr)]
    return res

data_clean = iqr_clean(data)
data_clean.shape
```

## Нормализация

```

data_norm = data_clean.copy()
for c in data_norm.select_dtypes(include=[np.number]).columns:
    data_norm[c]=(data_norm[c]-data_norm[c].mean())/data_norm[c].std()
data_norm.describe()
```
