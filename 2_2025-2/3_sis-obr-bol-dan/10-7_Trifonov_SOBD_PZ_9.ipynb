**Big Data** — это большие объёмы данных, характеризующиеся высокой скоростью поступления, разнообразием и требующие распределённых технологий хранения и обработки.

**Data pipeline** — это последовательность этапов обработки данных от их получения из источников до формирования конечного результата.

**Источник данных** — это система или сервис, из которого поступают данные (например, база данных, API, лог-файлы).

**Хранилище данных** — это система, предназначенная для долговременного хранения данных (например, базы данных, Data Lake).

**Вычислительный слой** — уровень системы, на котором выполняется обработка данных: очистка, преобразование, агрегация.

**Online-аналитика** — обработка данных в реальном времени или с минимальной задержкой.

**Оркестрация** — управление последовательностью выполнения задач, их зависимостями и расписанием.

**Data Engineer** — специалист, отвечающий за сбор, хранение, обработку данных и построение пайплайна.

## 7. Архитектурные слои системы
_Поясните, почему система разбита именно на такие слои, какой слой является наиболее важным и где возможны узкие места._

Система разбита на архитектурные слои для обеспечения модульности, масштабируемости и удобства сопровождения. Такое разделение позволяет изолировать различные этапы обработки данных и упростить развитие системы при увеличении объёма данных или изменении требований.

Слой источников данных отвечает за генерацию информации из различных каналов (тикеты, email, чат-боты, системы мониторинга). Слой интеграции обеспечивает сбор и передачу данных в систему, включая потоковую обработку. Слой хранения предназначен для централизованного накопления как сырых, так и обработанных данных.

Вычислительный слой выполняет ключевые операции по очистке, преобразованию и подготовке данных, а аналитический слой реализует модели машинного обучения для классификации заявок и принятия решений. Слой представления обеспечивает доступ пользователей к результатам системы, а слой оркестрации управляет выполнением всех процессов и контролирует их корректность.

Наиболее важным является **аналитический слой**, так как именно он определяет качество автоматической маршрутизации заявок. Ошибки на этом уровне напрямую влияют на бизнес-результат системы.

К возможным узким местам относятся:

* слой интеграции данных — из-за высокой нагрузки и необходимости обработки потоков в реальном времени
* вычислительный слой — при обработке больших объёмов неструктурированных данных (например, текстов)
* аналитический слой — из-за возможных ошибок моделей и необходимости их регулярного обновления

Таким образом, разделение на слои позволяет повысить устойчивость системы и упростить её масштабирование, но требует особого внимания к производительности и качеству обработки на критических этапах.

## 8. Архитектурная схема

Проектируемая система построена по многоуровневой архитектуре обработки данных. Данные поступают из различных источников: системы Service Desk, email-каналов, чат-ботов и систем мониторинга.

На вычислительном слое выполняется очистка, фильтрация и преобразование данных, включая обработку текстовой информации. После этого данные передаются в аналитический слой, где применяются модели машинного обучения для классификации заявок, определения их категории и приоритета.

## 15. Контрольные вопросы
1. Почему отдельная аналитическая модель не является промышленной системой сама по себе?  
_Ответ._ Аналитическая модель решает только задачу предсказания или классификации, но не обеспечивает полный жизненный цикл данных. Промышленная система включает сбор, хранение, обработку, интеграцию, автоматизацию и мониторинг, поэтому модель является лишь одним из её компонентов.

2. Что такое data pipeline и какие этапы он обычно включает?  
_Ответ._ Data pipeline — это последовательность этапов обработки данных от источников до конечного результата. Обычно он включает сбор данных, загрузку, очистку, преобразование, интеграцию, хранение, анализ и передачу результата пользователю или системе.

3. Чем offline-аналитика отличается от online-аналитики?  
_Ответ._ Offline-аналитика выполняется пакетно на накопленных данных с задержкой (например, ежедневно), а online-аналитика работает в реальном времени или с минимальной задержкой, обрабатывая данные сразу после их поступления.

4. Зачем нужна оркестрация задач обработки данных?  
_Ответ._ Оркестрация нужна для управления порядком выполнения задач, контроля зависимостей, автоматического запуска процессов по расписанию или событию, а также для обработки ошибок и повторного запуска этапов пайплайна.

5. В чём состоит различие между ролями Data Engineer и Data Scientist?  
_Ответ._ Data Engineer отвечает за сбор, хранение, интеграцию и поддержку работы пайплайна данных. Data Scientist занимается анализом данных, построением моделей машинного обучения, извлечением признаков и интерпретацией результатов.
